Features of Python
1. Object Oriented Programming - we can implement oop concepts like Polymorphism, Inheritance, Encapsulation, Abstraction.
2. Dynamically type language - we don't need to declare the data type of a variable explicitly.
3. Interpreted Language - executes line by line
4. Multi-Paridigm - supports multiple programming styles:
    object oriented programming = organize code and classes
    functional programming = organize the program as sequence of functions [Apply functions to data]
    imperative programming = variables, literals, algorithm to tell how it is to done
5. Extensible - can interact with multiple programming languages such as C/C++
6. Embeded - Python can be embedded inside applications written in other languages such as C++


Write a function to check if a string is a palindrome without using built-in string reversal methods like [::-1] or reversed()

```python
# Logic left read and right read should be same
def is_palindrome(s):
    # Normalized string
    normalized_str = ''.join(char.lower() for char in s if char.isalnum()) # already ignoring spaces and punctuations
    left = 0
    right = len(normalized_str) - 1
    
    while left < right:
        if normalized_str[left] != normalized_str[right]:
            return False
        left += 1
        right -= 1
    return True 

# print(is_palindrome("A man, a plan, a canal: Panama")) 
# print("Aman".isalnum()) # True
# print("Aman".isalpha()) # True
print("½".isnumeric()) # True
# print("123.07".isdecimal()) # False isnumeric() and isdecimal() are different. isdecimal() returns True only for decimal characters, while isnumeric() returns True for numeric characters, including fractions and subscripts.
# print("123".isdigit()) # True
```
