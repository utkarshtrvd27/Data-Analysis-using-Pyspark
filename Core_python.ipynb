Write a function to check if a string is a palindrome without using built-in string reversal methods like [::-1] or reversed()

```python
# Logic left read and right read should be same
def is_palindrome(s):
    # Normalized string
    normalized_str = ''.join(char.lower() for char in s if char.isalnum()) # already ignoring spaces and punctuations
    left = 0
    right = len(normalized_str) - 1
    
    while left < right:
        if normalized_str[left] != normalized_str[right]:
            return False
        left += 1
        right -= 1
    return True 

# print(is_palindrome("A man, a plan, a canal: Panama")) 
# print("Aman".isalnum()) # True
# print("Aman".isalpha()) # True
print("½".isnumeric()) # True
# print("123.07".isdecimal()) # False isnumeric() and isdecimal() are different. isdecimal() returns True only for decimal characters, while isnumeric() returns True for numeric characters, including fractions and subscripts.
# print("123".isdigit()) # True
```

### Functional Programming & Memory-Efficient Iteration

Iterator vs Generator
Iterator is an object that knows how to give you the next value.

Generator is a special type of iterator created using a function containing the yield keyword. Generator is more memory efficient.

```python
# Iterator example
numbers = [10, 20, 30]
iter_obj = iter(numbers) # numbers.__iter__()

print(next(iter_obj))  # 10 # iter_obj.__next__()
print(next(iter_obj))  # 20
print(next(iter_obj))  # 30
# print(next(iter_obj))  # StopIteration exception
```

```python
# Generator example
def my_generator():
    yield 1
    yield 2
    yield 3

        
my_gen = my_generator()
print(next(my_gen))  # 1
print(next(my_gen))  # 2
print(next(my_gen))  # 3
# print(next(my_gen))  # StopIteration exception
```

Implement a custom generator that streams a massive file line-by-line and yields only lines containing a specific keyword. Why is this critical for ETL?

The Concept: Traditional file parsing (.readlines()) forces the entire file into RAM, crashing systems on huge files. Generators use lazy evaluation, reading and yielding one line at a time while keeping memory footprint constant near 0 MB.

```python
def filter_log_stream(file_path: str, keyword: str):
    with open(file_path, "r", encoding="utf-8") as file:
        for line in file:
            if keyword in line:
                yield line.strip()
```

```python
gen = filter_log_stream("huge_sample_logs.txt", "ERROR")
# print(len(list(gen)))  # prints total number of lines containing "ERROR". You cannot use the generator again after this because it has been exhausted.

for i in range(5):
    print(next(gen))  # prints first 5 lines containing "ERROR"
```
